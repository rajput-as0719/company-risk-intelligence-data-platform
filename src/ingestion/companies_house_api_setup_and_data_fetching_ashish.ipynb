```python
import requests
import json
from requests.auth import HTTPBasicAuth
import pandas as pd
from datetime import datetime

# Companies House API Base URL
BASE_URL = "https://api.company-information.service.gov.uk"

# An API key needed from: https://developer.company-information.service.gov.uk/
# The API key is used as the username with no password
API_KEY = "d6bb76a0-3fdc-4736-b363-23e7e581bc11"  # 'your_api_key_here' -> Replace with your actual API key

# Authentication (API key as username, empty password)
auth = HTTPBasicAuth(API_KEY, '')

# Available endpoints and their purposes:
endpoints = {
    "company_profile": "/company/{company_number}",
    "company_officers": "/company/{company_number}/officers",
    "company_filing_history": "/company/{company_number}/filing-history",
    "company_charges": "/company/{company_number}/charges",
    "company_insolvency": "/company/{company_number}/insolvency",
    "company_persons_with_significant_control": "/company/{company_number}/persons-with-significant-control",
    "company_registers": "/company/{company_number}/registers",
    "search_companies": "/search/companies",
    "search_officers": "/search/officers",
    "search_disqualified_officers": "/search/disqualified-officers",
    "advanced_search": "/advanced-search/companies"
}

print("Companies House API Integration Setup")
print("="*50)
print(f"Base URL: {BASE_URL}")
print(f"\nAvailable endpoints:")
for name, endpoint in endpoints.items():
    print(f" - {name}: {endpoint}")
# Alternative way
'''for key in endpoints:
    print(f"  - {key}: {endpoints[key]}")
'''

# Function to search for companies
def search_companies(query, items_per_page=20, start_index=0):
    """
    Search for companies by name or number
    """
    url = f"{BASE_URL}/search/companies"
    params = {
        'q': query,
        'items_per_page': items_per_page,
        'start_index': start_index
    }
    
    response = requests.get(url, auth=auth, params=params)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None


# Function to get company profile
def get_company_profile(company_number):
    """
    Get detailed company profile information
    """
    url = f"{BASE_URL}/company/{company_number}"
    response = requests.get(url, auth=auth)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None


# Function to get company officers
def get_company_officers(company_number, items_per_page=35, start_index=0):
    """
    Get list of company officers (directors, secretaries, etc.)
    """
    url = f"{BASE_URL}/company/{company_number}/officers"
    params = {
        'items_per_page': items_per_page,
        'start_index': start_index
    }
    
    response = requests.get(url, auth=auth, params=params)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None


# Function to get filing history
def get_filing_history(company_number, items_per_page=25, start_index=0):
    """
    Get company filing history
    """
    url = f"{BASE_URL}/company/{company_number}/filing-history"
    params = {
        'items_per_page': items_per_page,
        'start_index': start_index
    }
    
    response = requests.get(url, auth=auth, params=params)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None


# Function to get persons with significant control (PSC)
def get_psc(company_number, items_per_page=25, start_index=0):
    """
    Get persons with significant control
    """
    url = f"{BASE_URL}/company/{company_number}/persons-with-significant-control"
    params = {
        'items_per_page': items_per_page,
        'start_index': start_index
    }
    
    response = requests.get(url, auth=auth, params=params)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None


# Function to get all data for a company
def get_all_company_data(company_number):
    """
    Fetch all available data for a given company
    """
    company_data = {
        'company_number': company_number,
        'fetched_at': datetime.now().isoformat(),
        'profile': None,
        'officers': None,
        'filing_history': None,
        'psc': None
    }
    
    print(f"Fetching data for company: {company_number}")
    
    # Get company profile
    print("  - Fetching company profile...")
    company_data['profile'] = get_company_profile(company_number)
    
    # Get officers
    print("  - Fetching officers...")
    company_data['officers'] = get_company_officers(company_number)
    
    # Get filing history
    print("  - Fetching filing history...")
    company_data['filing_history'] = get_filing_history(company_number)
    
    # Get PSC
    print("  - Fetching persons with significant control...")
    company_data['psc'] = get_psc(company_number)
    
    return company_data


print("\n" + "="*50)
print("Functions defined:")
print("  - search_companies(query)")
print("  - get_company_profile(company_number)")
print("  - get_company_officers(company_number)")
print("  - get_filing_history(company_number)")
print("  - get_psc(company_number)")
print("  - get_all_company_data(company_number)")
print("\nNext steps:")
print("  1. Get your API key from: https://developer.company-information.service.gov.uk/")
print("  2. Replace 'your_api_key_here' with your actual API key")
print("  3. Use the functions to fetch company data")
print("\nExample usage:")
print("  results = search_companies('Microsoft')")
print("  company_data = get_all_company_data('00000006')")
```

### Delta Format explained?

Delta format (Delta Lake) is an open-source storage layer that brings ACID transactions and reliability to data lakes built on cloud object storage like S3, ADLS, or GCS.

Key features:

ACID transactions — ensures data consistency even with concurrent reads/writes
Time travel — query previous versions of data using VERSION AS OF or TIMESTAMP AS OF
Schema enforcement — prevents bad data from being written by validating schema on write
Unified batch and streaming — same table can be read/written as both batch and streaming data
Optimized performance — uses Parquet file format with statistics, data skipping, and Z-ordering for faster queries
Audit history — maintains a transaction log (_delta_log/) tracking all changes

spark_df.write.format("delta").mode("overwrite").save(path) ->
This writes your Spark DataFrame as Delta files to the Volume path. The data is stored as Parquet files with metadata in _delta_log/. Unlike plain JSON or Parquet, Delta tables can be updated, merged, deleted, and versioned.

Why use Delta over JSON:

JSON is human-readable but slow for analytics and doesn't support updates
Delta is optimized for queries, supports updates/deletes, and provides data versioning
Delta integrates natively with Unity Catalog for governance
